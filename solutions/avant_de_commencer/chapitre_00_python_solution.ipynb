### Exercice 1 · Variables et types — niveau ●

Une variable est une étiquette collée sur une valeur. Ici, une chaîne de
caractères (`str`). Deux outils : la fonction `len(...)` donne la longueur,
la méthode `.lower()` renvoie la version en minuscules.

```python
titre = "LA CIGALE ET LA FOURMI"

nb = len(titre)
minuscules = titre.lower()
```

```python
# Validation : variables et types.
assert nb == 22, f"nb devrait valoir 22, pas {nb}"
assert minuscules == "la cigale et la fourmi", f"minuscules obtenu : {minuscules!r}"
print("Variables OK :", nb, "caractères,", repr(minuscules))
```

### Exercice 2 · Listes et slicing — niveau ●

Les indices partent de 0, les indices négatifs comptent depuis la fin, et une
tranche `liste[debut:fin]` inclut la borne de gauche mais exclut celle de droite.

```python
vers = [
    "La cigale, ayant chanté",
    "Tout l'été,",
    "Se trouva fort dépourvue",
    "Quand la bise fut venue :",
]

premier = vers[0]
dernier = vers[-1]
milieu = vers[1:3]
```

```python
# Validation : listes et slicing.
assert premier == "La cigale, ayant chanté", f"premier obtenu : {premier!r}"
assert dernier == "Quand la bise fut venue :", f"dernier obtenu : {dernier!r}"
assert milieu == ["Tout l'été,", "Se trouva fort dépourvue"], f"milieu obtenu : {milieu!r}"
print("Listes OK : les trois découpes sont justes.")
```

### Exercice 4 · Référence vs copie — niveau ●●

`b = a` ne copie pas une liste : les deux noms désignent la même liste.
Pour une copie indépendante : `list(a)` ou `a[:]`.

```python
original = [4, 8, 15]

copie = list(original)
copie.append(16)
```

```python
# Validation : référence vs copie.
assert copie == [4, 8, 15, 16], f"copie obtenu : {copie!r}"
assert original == [4, 8, 15], f"original a changé : {original!r} (tu as copié la référence ?)"
assert copie is not original, "copie et original désignent la même liste"
print("Copie OK : original est intact.")
```

### Exercice 5 · Dictionnaires — niveau ●●

La table de correspondance caractère → numéro du chapitre 1, en miniature.
Boucle `for` avec `enumerate`, ou compréhension de dictionnaire : au choix.

```python
chars = ["a", "b", "c", "d"]

stoi = {c: i for i, c in enumerate(chars)}
```

```python
# Validation : dictionnaires.
assert stoi == {"a": 0, "b": 1, "c": 2, "d": 3}, f"stoi obtenu : {stoi!r}"
assert stoi.get("z", -1) == -1, "get doit renvoyer la valeur de repli pour une clé absente"
print("Dictionnaire OK :", stoi)
```

### Exercice 6 · Boucles — niveau ●●

Une boucle `for` parcourt une collection ; un `if` filtre au passage.
Compte les « r » de la phrase, sans utiliser la méthode `.count()`.

```python
phrase = "maître corbeau sur un arbre perché"

nb_r = 0
for c in phrase:
    if c == "r":
        nb_r = nb_r + 1
```

```python
# Validation : boucles.
assert nb_r == phrase.count("r"), f"nb_r devrait valoir {phrase.count('r')}, pas {nb_r}"
print(f"Boucle OK : {nb_r} lettres r.")
```

### Exercice 7 · Compréhensions de listes — niveau ●●

Une boucle qui transforme une liste en une autre tient en une ligne :
`[expression for element in liste]`, avec un `if` optionnel en queue.

```python
mots = ["le", "corbeau", "et", "le", "renard"]

tailles = [len(m) for m in mots]
longs = [m for m in mots if len(m) > 2]
```

```python
# Validation : compréhensions.
assert tailles == [2, 7, 2, 2, 6], f"tailles obtenu : {tailles!r}"
assert longs == ["corbeau", "renard"], f"longs obtenu : {longs!r}"
print("Compréhensions OK :", tailles, "|", longs)
```

### Exercice 8 · Fonctions et f-strings — niveau ●●●

Une fonction avec une valeur par défaut, qui fabrique sa réponse avec une
f-string (le `f` devant les guillemets, les accolades autour des valeurs).

```python
def presenter(mot, n=1):
    return f"{mot} apparaît {n} fois"
```

```python
# Validation : fonctions et f-strings.
assert presenter("loup", 3) == "loup apparaît 3 fois", f"obtenu : {presenter('loup', 3)!r}"
assert presenter("agneau") == "agneau apparaît 1 fois", f"obtenu : {presenter('agneau')!r}"
print("Fonction OK :", presenter("fourmi", 2))
```

### Exercice 9 · Une classe minimale — niveau ●●●

Des données (les attributs) plus une notice (les méthodes). C'est la forme
exacte de la classe `Value` que tu écriras au chapitre 4 pour ton moteur
d'autograd : plus riche, mais sur ce squelette.

```python
class Compteur:
    def __init__(self):
        self.total = 0

    def ajouter(self, n):
        self.total = self.total + n

    def __repr__(self):
        return f"Compteur(total={self.total})"
```

```python
# Validation : classe minimale.
c = Compteur()
c.ajouter(3)
c.ajouter(4)
assert c.total == 7, f"c.total devrait valoir 7, pas {c.total}"
assert repr(c) == "Compteur(total=7)", f"__repr__ obtenu : {repr(c)!r}"
c2 = Compteur()
assert c2.total == 0, "chaque objet doit avoir SON total, qui démarre à 0"
print("Classe OK :", c, "|", c2)
```
