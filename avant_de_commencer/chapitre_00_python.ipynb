### 3.1 Le slicing : découper une tranche

`texte[debut:fin]`. Règle unique, à graver : la borne de gauche est **incluse**,
la borne de droite est **exclue**.

```python
texte = "le loup"
print(texte[3:7])       # "loup" : indices 3, 4, 5, 6
print(texte[:2])        # "le" : borne gauche absente = depuis le début
print(texte[3:])        # "loup" : borne droite absente = jusqu'au bout
print(texte[-4:])       # "loup" : les 4 derniers caractères
```

### 3.2 Le piège : deux noms, une seule liste

`b = a` ne copie **pas** la liste : les deux noms désignent la même liste en mémoire.
Pour une copie indépendante, demande-la explicitement : `list(a)` ou `a[:]`.

```python
a = [5, 3, 8]
b = a               # b et a : la même liste
b.append(1)
print(a)            # [5, 3, 8, 1]  <- a aussi a changé !

c = list(a)         # list(a) ou a[:] : une vraie copie
c.append(2)
print(a)            # [5, 3, 8, 1]  <- a n'a pas bougé
```

## 5. Boucles : répéter sans se lasser

La boucle `for` répète un bloc pour chaque élément. Le bloc répété est signalé par
l'**indentation** : les quatre espaces font partie de la grammaire, pas de la décoration.

```python
for mot in ["le", "loup", "et"]:
    print(mot)

for i in range(3):       # range(3) produit 0, 1, 2
    print(i)

for i, c in enumerate("abc"):    # l'indice ET la valeur
    print(i, c)                  # 0 a, puis 1 b, puis 2 c
```

## 6. Fonctions : donner un nom à un geste

Une fonction emballe un bloc de code sous un nom, avec des entrées (les **arguments**)
et une sortie (le `return`).

```python
def compter(texte, lettres):
    total = 0
    for c in texte:
        if c in lettres:
            total = total + 1
    return total

n = compter("le loup", "aeiou")      # n vaut 3
print(n)
```

Un argument peut avoir une **valeur par défaut**, utilisée quand l'appelant ne précise rien :

```python
def generer(prompt="\n", longueur=300):
    ...

generer()                        # prompt "\n", longueur 300
generer("Le loup")               # prompt "Le loup", longueur 300
generer("Le loup", longueur=50)  # argument nommé : plus lisible
```

Dernier geste, qui surprend toujours : en Python, **une fonction est une valeur comme une autre**. On peut la ranger dans une variable et surtout la **passer en argument** à une autre fonction. Et pour une toute petite fonction d'une ligne, `lambda` en fabrique une sans passer par `def`. Tu reverras ce geste dès le chapitre 3 : le capteur de pente `pente(f, x)` reçoit une fonction en argument.

```python
def appliquer(f, valeur):
    return f(valeur)                  # f est une fonction reçue en argument

carre = lambda x: x ** 2              # même chose que def carre(x): return x ** 2
print(appliquer(carre, 9))            # 81
print(appliquer(len, "le loup"))      # 7 : len se passe comme n'importe quelle valeur
```

## 7. f-strings : fabriquer du texte propre

Un `f` devant les guillemets, des accolades autour de ce qu'il faut calculer.
Après les deux-points, le **format** : `:.2f` arrondit l'affichage à deux décimales,
`:5d` cale l'entier sur cinq caractères.

```python
mot = "loup"
n = 8
print(f"le mot {mot} a {len(mot)} lettres")   # le mot loup a 4 lettres
print(f"{n} voyelles")                        # 8 voyelles

loss = 4.4234891
step = 40
print(f"étape {step:5d} | loss {loss:.2f}")   # étape    40 | loss 4.42
```

## 8. Compréhensions de listes : la boucle en une ligne

« La liste des `len(m)` pour chaque `m` dans `mots` », avec un `if` optionnel en queue.
La même écriture existe pour les dictionnaires.

```python
mots = ["le", "loup", "et"]

# version boucle
tailles = []
for m in mots:
    tailles.append(len(m))

# version compréhension : le même résultat, en une ligne
tailles = [len(m) for m in mots]          # [2, 4, 2]
print(tailles)

longs = [m for m in mots if len(m) > 2]   # ["loup"]
print(longs)

chars = ["a", "b", "c"]
stoi = {c: i for i, c in enumerate(chars)}    # {"a": 0, "b": 1, "c": 2}
print(stoi)
```

## 9. Une classe minimale : des données et une notice

Une **classe** regroupe des données (les **attributs**) et les gestes qui les
manipulent (les **méthodes**). `__init__` construit l'objet, `self` désigne l'objet
en train d'agir, `__repr__` dit comment il s'affiche.

```python
class Compteur:
    def __init__(self):
        self.total = 0                    # un attribut : une donnée de l'objet

    def ajouter(self, n):
        self.total = self.total + n       # une méthode : un geste sur ces données

    def __repr__(self):
        return f"Compteur(total={self.total})"

c = Compteur()          # fabrique un objet : __init__ s'exécute
c.ajouter(3)
c.ajouter(4)
print(c.total)          # 7
print(c)                # Compteur(total=7) : c'est __repr__ qui parle
```
